```python
from anthropic import Anthropic

client = Anthropic()
model = "claude-sonnet-5"
```

```python
messages = client.messages.create(
    model=model,
    max_tokens=1000,
    messages=[
        {
            "role": "user",
            "content": "What is quantum computing? Answer in 1 sentence?"
        }
    ]
)
```
